# Optuna Dashboard & Live Sweeps in Jupyter Lab

This notebook provides two ways to monitor your live Optuna sweeps and view Pareto charts directly inside Jupyter Lab.

## 1. Quick In-Cell Visualization (Recommended)
You can plot the live Pareto front right here. Just re-run this cell whenever you want to see the latest progress.

````python
import optuna
from optuna.visualization import plot_pareto_front

# 1. Point to your SQLite database. Change the path to match your actively running study.
storage_url = "sqlite:///../results/optuna/master_study_dqn/master_study_dqn.db"

# 2. Load the study by name
study_name = "master_study_dqn"
study = optuna.load_study(study_name=study_name, storage=storage_url)

# 3. Plot the interactive live Pareto Front!
# If you have >3 objectives, you can target specific ones using the `target_names` parameter.
fig = plot_pareto_front(study)
fig.show()
````

## 2. Using the Full Optuna Dashboard in JupyterLab

If you want the full dashboard UI rather than just the chart:

1. Open a new terminal cell inside Jupyter and run:
   ```bash
   pip install jupyterlab jupyterlab-optuna optuna-dashboard
   ```
2. Refresh your JupyterLab page in the browser.
3. You will now see an **Optuna** tile in the JupyterLab launcher menu (where you normally open Python notebooks).
4. Click it, and paste your database URL (e.g., `sqlite:///results/optuna/master_study_dqn/master_study_dqn.db`) into the dialog box.

Alternatively, you can just run `optuna-dashboard sqlite:///../results/optuna/master_study_dqn/master_study_dqn.db` inside a terminal tab in Jupyter, and it will give you a local URL to click on.
